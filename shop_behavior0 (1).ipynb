```python
#import panda & numpy
import pandas as pd
import numpy as np
```

```python
#import  dataset into pandas & numpy liberary 'customer_shopp_behavior'
df = pd.read_csv('customer_shopp_behavior.csv')
```

```python
# Give overall infomation about dataset
df.info()
```

```python
# check null/missing values that are  present in dataset
df.isnull().sum()
```

```python
# remove null/missing values from the dataset with the help of median
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
df.isnull().sum()
```

```python
# 1.change the column name into lower case
# 2. replace ' ' into (_)
# 3. rename the columns (to  ease the further EDA )
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.head()
```

```python
# give the overall statetics summary of the data to know more about customer interests
df.describe(include='all')
```
